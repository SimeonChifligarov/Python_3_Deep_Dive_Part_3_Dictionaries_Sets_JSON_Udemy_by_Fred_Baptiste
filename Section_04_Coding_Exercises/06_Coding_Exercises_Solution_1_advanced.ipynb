# Exercise 1 — Sort a Dictionary by Value

## Objective

Write a Python function that creates and returns a new dictionary whose items are ordered by their values.

We assume:

- all dictionary values are mutually comparable;
- values have a natural sort order;
- the original dictionary should not be modified.

A Pythonic solution can combine `sorted()` with a dictionary comprehension.

## Example input

Given:

````python
composers = {
    'Johann': 65,
    'Ludwig': 56,
    'Frederic': 39,
    'Wolfgang': 35,
}
````

The desired ordering is:

```python
{
    'Wolfgang': 35,
    'Frederic': 39,
    'Ludwig': 56,
    'Johann': 65,
}
```

## Solution

`dict.items()` produces `(key, value)` pairs. We sort those pairs according to the second element — the value — and then construct a new dictionary from the sorted pairs.

Using a dictionary comprehension keeps the implementation concise and avoids incrementally building an initially empty dictionary.

````python
from typing import Dict, Hashable, Mapping, TypeVar


K = TypeVar('K', bound=Hashable)
V = TypeVar('V')


def sort_dict_by_value(mapping: Mapping[K, V], reverse: bool = False) -> Dict[K, V]:
    """Return a new dictionary ordered by its values.

    Parameters
    ----------
    mapping:
        Mapping whose values are mutually comparable.
    reverse:
        If False, sort from smallest to largest.
        If True, sort from largest to smallest.

    Returns
    -------
    dict
        A new dictionary containing the same key/value pairs in
        value-sorted order.

    Notes
    -----
    Python's sorting algorithm is stable. Therefore, items with equal
    values retain their relative order from the input mapping.

    The input mapping is never modified.
    """
    return {
        key: value
        for key, value in sorted(
            mapping.items(),
            key=lambda item: item[1],
            reverse=reverse,
        )
    }
````

## Run the function

````python
sorted_composers = sort_dict_by_value(composers)

print(sorted_composers)
````

Expected output:

```text
{'Wolfgang': 35, 'Frederic': 39, 'Ludwig': 56, 'Johann': 65}
```

## How it works

Consider:

```python
composers.items()
```

Conceptually, this gives pairs such as:

```python
('Johann', 65)
('Ludwig', 56)
('Frederic', 39)
('Wolfgang', 35)
```

The expression:

```python
key=lambda item: item[1]
```

tells `sorted()` to compare the second element of each pair, which is the dictionary value.

The resulting sorted pairs are then turned into a new dictionary by the dictionary comprehension.

````python
sorted_items = sorted(
    composers.items(),
    key=lambda item: item[1],
)

print(sorted_items)
````

Expected intermediate result:

```text
[('Wolfgang', 35), ('Frederic', 39), ('Ludwig', 56), ('Johann', 65)]
```

## Verification

Use assertions to verify both the result and the fact that the original dictionary remains unchanged.

````python
expected = {
    'Wolfgang': 35,
    'Frederic': 39,
    'Ludwig': 56,
    'Johann': 65,
}

original = composers.copy()
result = sort_dict_by_value(composers)

assert result == expected
assert list(result.items()) == list(expected.items())
assert composers == original
assert result is not composers

print('All basic checks passed.')
````

## Additional tests

A robust implementation should also behave correctly for empty dictionaries, single-item dictionaries, negative values, duplicate values, and descending sorting.

````python
# Empty mapping
assert sort_dict_by_value({}) == {}

# Single item
assert sort_dict_by_value({'a': 10}) == {'a': 10}

# Negative and positive values
numbers = {'a': 4, 'b': -3, 'c': 0, 'd': -10}
assert list(sort_dict_by_value(numbers).items()) == [
    ('d', -10),
    ('b', -3),
    ('c', 0),
    ('a', 4),
]

# Descending order
assert list(sort_dict_by_value(numbers, reverse=True).items()) == [
    ('a', 4),
    ('c', 0),
    ('b', -3),
    ('d', -10),
]

# Equal values: stable sorting preserves their original relative order
ties = {'first': 10, 'second': 5, 'third': 10, 'fourth': 5}
assert list(sort_dict_by_value(ties).items()) == [
    ('second', 5),
    ('fourth', 5),
    ('first', 10),
    ('third', 10),
]

print('All edge-case checks passed.')
````

## Alternative concise implementation

If only the exact requirements of the exercise are needed, the function can be reduced to one expression:

````python
def sort_dict_by_value_simple(d):
    return {key: value for key, value in sorted(d.items(), key=lambda item: item[1])}


print(sort_dict_by_value_simple(composers))
````

The more fully documented `sort_dict_by_value()` version is preferable in reusable application code, while the concise version expresses the core solution particularly clearly.

## Complexity analysis

Let **n** be the number of dictionary items.

- `mapping.items()` exposes the **n** key/value pairs.
- `sorted()` requires **O(n log n)** time in the general case.
- constructing the new dictionary requires **O(n)** time.
- the sorted list and resulting dictionary require **O(n)** additional space.

Therefore:

**Time complexity:** `O(n log n)`

**Auxiliary space:** `O(n)`

## Important note about dictionary ordering

The solution relies on dictionaries preserving insertion order when iterated.

Insertion ordering is guaranteed by the Python language specification in **Python 3.7+**. CPython 3.6 also preserves insertion order as an implementation detail, but Python 3.7 or newer should be preferred when dictionary ordering is part of program behavior.

## Final answer

The essential Pythonic solution is:

```python
def sort_dict_by_value(d):
    return {
        key: value
        for key, value in sorted(d.items(), key=lambda item: item[1])
    }
```

This solution is concise, does not mutate its input, uses Python's built-in sorting facilities, and constructs the result directly rather than building an empty dictionary one item at a time.
